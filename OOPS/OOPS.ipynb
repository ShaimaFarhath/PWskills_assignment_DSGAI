```python
#2. Python Class for a Car
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print(f"Car: {self.year} {self.make} {self.model}")

# Example usage
car1 = Car("Toyota", "Camry", 2022)
car1.display_info()

```

```python
'''3. Difference between Instance Methods and Class Methods
•	Instance Methods: Operate on an instance of the class and can access instance attributes (self).
•	Class Methods: Operate on the class itself and use @classmethod with cls as the first parameter.
Example:
class Example:
    count = 0  # Class attribute

    def __init__(self, value):
        self.value = value  # Instance attribute

    def instance_method(self):
        return f"Instance method: {self.value}"

    @classmethod
    def class_method(cls):
    
        return f"Class method: {cls.count}"

obj = Example(10)
print(obj.instance_method())  # Works on object
print(Example.class_method())  # Works on class
```

```python
'''Q 4,5,6,7
4. Method Overloading in Python
Python does not support method overloading like Java/C++. Instead, it can be handled using default parameters or *args:
class MathOperations:
    def add(self, a, b, c=0):
        return a + b + c

math_obj = MathOperations()
print(math_obj.add(2, 3))  # Uses two parameters
print(math_obj.add(2, 3, 4))  # Uses three parameters

5. Three Access Modifiers in Python
1.	Public (name) – Accessible everywhere.
2.	Protected (_name) – Accessible within the class and subclasses.
3.	Private (__name) – Not directly accessible outside the class.
Example:
class Example:
    def __init__(self):
        self.public = "Public"
        self._protected = "Protected"
        self.__private = "Private"

obj = Example()
print(obj.public)        # ✅ Accessible
print(obj._protected)    # ✅ Accessible (but should not be)
# print(obj.__private)   # ❌ Not accessible (Name Mangling)

6. Five Types of Inheritance & Multiple Inheritance Example
1.	Single Inheritance
2.	Multiple Inheritance
3.	Multilevel Inheritance
4.	Hierarchical Inheritance
5.	Hybrid Inheritance
Example of Multiple Inheritance:
class A:
    def method_a(self):
        print("Method A")

class B:
    def method_b(self):
        print("Method B")

class C(A, B):  # Inherits from both A and B
    pass

obj = C()
obj.method_a()
obj.method_b()

7. Method Resolution Order (MRO)
•	MRO is the order in which Python looks for a method in multiple inherited classes.
•	Retrieve it using:
print(C.mro())

```

```python
#8. Abstract Base Class
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# Example Usage
c = Circle(5)
r = Rectangle(4, 6)
print(c.area())
print(r.area())
```

```python
#9. Polymorphism Example
def print_area(shape):
    print(f"Area: {shape.area()}")

print_area(Circle(5))
print_area(Rectangle(4, 6))

#10. Encapsulation in BankAccount Class
class BankAccount:
    def __init__(self, account_number, balance):
        self.__account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds")

    def get_balance(self):
        return self.__balance

# Example Usage
acc = BankAccount(12345, 1000)
acc.deposit(500)
print(acc.get_balance())  # Output: 1500

#11. Overriding __str__ and __add__
class Number:
    def __init__(self, value):
        self.value = value

    def __str__(self):
        return f"Number: {self.value}"

    def __add__(self, other):
        return Number(self.value + other.value)

num1 = Number(10)
num2 = Number(20)
print(num1)  # Calls __str__
print(num1 + num2)  # Calls __add__

#12. Execution Time Decorator
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Execution Time: {end - start:.4f} seconds")
        return result
    return wrapper

@timer
def slow_function():
    time.sleep(2)
    print("Function completed")

slow_function()

#13. Diamond Problem & Python Resolution
#•	Occurs when a class inherits from multiple classes that have a common ancestor.
#•	Resolved using MRO in Python (super()).
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")

class C(A):
    def show(self):
        print("C")

class D(B, C):
    pass

obj = D()
obj.show()  # Follows MRO: B → C → A
print(D.mro())  # Prints MRO

#14. Class Method to Track Instances
class Counter:
    count = 0

    def __init__(self):
        Counter.count += 1

    @classmethod
    def instance_count(cls):
        return f"Instances created: {cls.count}"

obj1 = Counter()
obj2 = Counter()
print(Counter.instance_count())

#15. Static Method to Check Leap Year
class Utility:
    @staticmethod
    def is_leap_year(year):
        return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

print(Utility.is_leap_year(2024))  # True


```
