```python
#1. Bokeh Plot of a Sine Wave
from bokeh.plotting import figure, show
import numpy as np

# Generate data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create a figure
p = figure(title="Sine Wave", x_axis_label='X', y_axis_label='sin(X)', width=700, height=400)

# Add line
p.line(x, y, line_width=2, color='blue')

show(p)
```

```python
#2. Bokeh Scatter Plot with Random Data
import pandas as pd
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.models import ColumnDataSource
import numpy as np

output_notebook()

# Generate random data
np.random.seed(42)
data = {
    'x': np.random.rand(50) * 10,
    'y': np.random.rand(50) * 10,
    'sizes': np.random.randint(10, 50, 50),
    'colors': np.random.randint(50, 255, 50)
}

df = pd.DataFrame(data)

# Create figure
p = figure(title="Random Scatter Plot", width=700, height=400)

# Scatter plot
p.scatter(df['x'], df['y'], size=df['sizes'], color=df['colors'], alpha=0.6)

show(p)
```

```python
#3. Bokeh Bar Chart for Fruit Counts
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource
from bokeh.io import output_notebook

output_notebook()

# Data
fruits = ['Apples', 'Oranges', 'Bananas', 'Pears']
counts = [20, 25, 30, 35]

# Create figure
p = figure(x_range=fruits, title="Fruit Counts", width=700, height=400)
p.vbar(x=fruits, top=counts, width=0.5, color=['red', 'orange', 'yellow', 'green'])

show(p)
```

```python
#4. Bokeh Histogram
from bokeh.plotting import figure, show
import numpy as np

# Generate random data
data_hist = np.random.randn(1000)

# Create histogram bins
hist, edges = np.histogram(data_hist, bins=30)

# Create figure
p = figure(title="Histogram of Random Data", width=700, height=400)
p.quad(top=hist, bottom=0, left=edges[:-1], right=edges[1:], fill_color="navy", line_color="white")

show(p)
```

```python
#5. Bokeh Heatmap
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, LinearColorMapper, ColorBar
from bokeh.transform import transform
import numpy as np

# Sample data
x = np.linspace(0, 1, 10)
y = np.linspace(0, 1, 10)
x_grid, y_grid = np.meshgrid(x, y)
z_grid = np.sin(x_grid * np.pi) * np.cos(y_grid * np.pi)

# Flatten the arrays
x_flat, y_flat, z_flat = x_grid.flatten(), y_grid.flatten(), z_grid.flatten()

# Create a ColumnDataSource
source = ColumnDataSource(data={'x': x_flat, 'y': y_flat, 'z': z_flat})

# Create figure
p = figure(title="Heatmap", x_axis_label="X", y_axis_label="Y")

# Define color mapper
mapper = LinearColorMapper(palette="Viridis256", low=min(z_flat), high=max(z_flat))

# Draw rectangles using ColumnDataSource
p.rect(x='x', y='y', width=0.1, height=0.1, fill_color=transform('z', mapper), line_color=None, source=source)

# Add color bar
color_bar = ColorBar(color_mapper=mapper, location=(0, 0))
p.add_layout(color_bar, 'right')

show(p)
```
